```python
from typing import Optional
import numpy as np
import gymnasium as gym

class ScheduleEnv(gym.Env):
    def __init__(self, num_employees: int = 10, num_days: int = 31, num_shifts: int = 3):
        super(ScheduleEnv, self).__init__()
        self.num_employees = num_employees
        self.num_days = num_days
        self.num_shifts = num_shifts
        self.minimum_coverage = 3

        self.total_steps = num_employees * num_days

        # Features: [Employee ID, Day ID]
        self.num_features = 2

        self.max_days_per_month = 22
        self.max_consecutive_days = 5

        matrix_height = self.total_steps
        matrix_width = self.num_features + self.num_shifts
        self.observation_space = gym.spaces.Box(
            low=0,
            high=max(num_employees, num_days),
            shape=(matrix_height, matrix_width),
            dtype=np.float32
        )

        self.action_space = gym.spaces.Discrete(num_shifts)

        self.reset()


    def _build_initial_matrix(self):
        matrix = np.zeros((self.total_steps, self.num_features + self.num_shifts), dtype=np.float32)
        row_idx = 0
        for emp in range(self.num_employees):
            for day in range(self.num_days):
                matrix[row_idx, 0] = emp  
                matrix[row_idx, 1] = day  
                row_idx += 1
        return matrix
    
    def _get_info(self):
        # TODO: Implement
        return {}
    
    def _calculate_reward(self, emp_id: int, day_id: int, action: int):
        reward = 0

        if action == 0:
            if self.consecutive_days[emp_id] >= self.max_consecutive_days:
                reward += 5
        else:
            if action == 1 and self.daily_coverage_M[day_id] < self.minimum_coverage:
                reward += 4
            elif action == 2 and self.daily_coverage_T[day_id] < self.minimum_coverage:
                reward += 4

            if action == 1 and self.daily_coverage_M[day_id] >= self.minimum_coverage:
                reward -= 2
            elif action == 2 and self.daily_coverage_T[day_id] >= self.minimum_coverage:
                reward -= 2

        return reward
    
    def _calculate_final_reward(self):
        reward = 0

        violations = 0
        for day in range(self.num_days):
            if self.daily_coverage_M[day] < self.minimum_coverage:
                reward -= (self.minimum_coverage - self.daily_coverage_M[day]) * 10
                violations += 1
            if self.daily_coverage_T[day] < self.minimum_coverage:
                reward -= (self.minimum_coverage - self.daily_coverage_T[day]) * 10
                violations += 1

        if violations == 0:
            reward += 100

        return reward
    
    def reset(self, seed: Optional[int] = None, options: Optional[dict] = None):
        super().reset(seed=seed)
        self.state = self._build_initial_matrix()
        self.current_step = 0

        self.days_worked = np.zeros(self.num_employees, dtype=int)
        self.consecutive_days = np.zeros(self.num_employees, dtype=int)

        self.daily_coverage_T = np.zeros(self.num_days, dtype=np.float32)
        self.daily_coverage_M = np.zeros(self.num_days, dtype=np.float32)

        return self.state, self._get_info()
    
    def step(self, action: int):
        emp_id = int(self.state[self.current_step, 0])
        day_id = int(self.state[self.current_step, 1])

        self.state[self.current_step, 2 + action] = 1

        if action == 0:
            self.consecutive_days[emp_id] = 0
        else:
            self.days_worked[emp_id] += 1
            self.consecutive_days[emp_id] += 1

            if action == 1:
                self.daily_coverage_M[day_id] += 1
            elif action == 2:
                self.daily_coverage_T[day_id] += 1

        reward = self._calculate_reward(emp_id, day_id, action)

        self.current_step += 1
        
        terminated = self.current_step >= self.total_steps
        if terminated:
            reward += self._calculate_final_reward()

        truncated = False

        return self.state.copy(), reward, terminated, truncated, self._get_info()
    
    def render(self):
        print("Current Schedule State:")
        for emp in range(self.num_employees):
            emp_schedule = self.state[emp*self.num_days:(emp+1)*self.num_days, 2:]
            print(f"Employee {emp}: {emp_schedule}")

    def get_action_mask(self):
        if self.current_step >= self.total_steps:
            return np.array([True, False, False])
            
        emp = int(self.state[self.current_step, 0])
        day = int(self.state[self.current_step, 1])
        
        mask = np.ones(3, dtype=bool)
        
        if self.days_worked[emp] >= self.max_days_per_month:
            mask[1] = False
            mask[2] = False
            
        if self.consecutive_days[emp] >= self.max_consecutive_days:
            mask[1] = False
            mask[2] = False
            
        if day > 0:
            yesterday_step = (emp * self.num_days) + (day - 1)
            worked_afternoon_yesterday = self.state[yesterday_step, 2 + 2] == 2
            if worked_afternoon_yesterday:
                mask[1] = False 
                
        return mask
```

```python
import dgl
import torch
import numpy as np

def convert_env_to_graph(env):
    sources = env.state[:, 0].astype(np.int64)     
    destinations = env.state[:, 1].astype(np.int64)
    
    graph_data = {
        ('employee', 'assigned_to', 'day'): (torch.tensor(sources), torch.tensor(destinations)),
        ('day', 'staffed_by', 'employee'): (torch.tensor(destinations), torch.tensor(sources))
    }
    g = dgl.heterograph(graph_data)
    
    emp_feat = np.column_stack((env.days_worked, env.consecutive_days))
    day_feat = np.column_stack((env.daily_coverage_M, env.daily_coverage_T))
    g.nodes['employee'].data['feat'] = torch.tensor(emp_feat, dtype=torch.float32)
    g.nodes['day'].data['feat'] = torch.tensor(day_feat, dtype=torch.float32)
    
    edge_attr = env.state[:, 2:5]
    g.edges['assigned_to'].data['feat'] = torch.tensor(edge_attr, dtype=torch.float32)
    
    return g
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import dgl.nn as dglnn

class PolicyNetwork(nn.Module):
    def __init__(self, emp_in_feats=2, day_in_feats=2, hidden_dim=16, encoded_dim=64, num_actions=3):
        super(PolicyNetwork, self).__init__()
        
        self.emp_proj = nn.Linear(emp_in_feats, hidden_dim)
        self.day_proj = nn.Linear(day_in_feats, hidden_dim)
        
        self.conv1 = dglnn.HeteroGraphConv({
            'assigned_to': dglnn.SAGEConv(hidden_dim, 32, 'mean'),
            'staffed_by': dglnn.SAGEConv(hidden_dim, 32, 'mean')
        }, aggregate='sum') 
        
        self.conv2 = dglnn.HeteroGraphConv({
            'assigned_to': dglnn.SAGEConv(32, encoded_dim, 'mean'),
            'staffed_by': dglnn.SAGEConv(32, encoded_dim, 'mean')
        }, aggregate='sum')
        
        self.scoring_layer = nn.Linear(encoded_dim * 2, num_actions)

    def forward(self, g, current_emp_id, current_day_id, action_mask):
        raw_emp_feat = g.nodes['employee'].data['feat']
        raw_day_feat = g.nodes['day'].data['feat']
        
        h = {
            'employee': F.relu(self.emp_proj(raw_emp_feat)),
            'day': F.relu(self.day_proj(raw_day_feat))
        }
        
        h = self.conv1(g, h)
        h = {k: F.relu(v) for k, v in h.items()}
        h = {k: F.dropout(v, p=0.5, training=self.training) for k, v in h.items()}
        
        h = self.conv2(g, h)
        
        emp_emb = h['employee'][current_emp_id]
        day_emb = h['day'][current_day_id]
        
        combined_state = torch.cat([emp_emb, day_emb], dim=-1)
        logits = self.scoring_layer(combined_state)
        
        mask_tensor = torch.tensor(action_mask, dtype=torch.bool)
        logits = logits.masked_fill(~mask_tensor, float('-inf'))
        
        log_probs = F.log_softmax(logits, dim=-1)
        
        return log_probs
```

```python
import torch
import torch.optim as optim
from torch.distributions import Categorical
import numpy as np

env = ScheduleEnv(num_employees=10, num_days=31)
policy_net = PolicyNetwork(emp_in_feats=2, day_in_feats=2, hidden_dim=16, encoded_dim=64, num_actions=3)

optimizer = optim.Adam(policy_net.parameters(), lr=1e-3)

num_episodes = 2000
baseline_reward = 0.0 # This will track the average reward to stabilize training
beta = 0.8 # Decay rate for the exponential moving average baseline

print("Starting GNN Training...")

for episode in range(num_episodes):
    env.reset()
    
    # Memory to store the agent's decisions and rewards for this episode
    saved_log_probs = []
    rewards = []
    
    terminated = False
    truncated = False
    
    while not (terminated or truncated):
        g = convert_env_to_graph(env)
        
        action_mask = env.get_action_mask()
        emp_id = int(env.state[env.current_step, 0])
        day_id = int(env.state[env.current_step, 1])
        
        log_probs = policy_net(g, emp_id, day_id, action_mask)
        probs = torch.exp(log_probs)
        
        # Sample an action based on the probabilities
        m = Categorical(probs)
        action = m.sample()
        
        _, reward, terminated, truncated, _ = env.step(action.item())
        
        # Save the log-probability and the reward to memory
        saved_log_probs.append(m.log_prob(action))
        rewards.append(reward)
        
    total_reward = sum(rewards)
    
    # Update the baseline (Exponential Moving Average) 
    if episode == 0:
        baseline_reward = total_reward
    else:
        baseline_reward = beta * baseline_reward + (1 - beta) * total_reward
        
    # Advantage: Did this schedule perform better or worse than average?
    advantage = total_reward - baseline_reward
    
    policy_loss = []
    for log_prob in saved_log_probs:
        # If advantage is positive, loss is negative (encourages action)
        # If advantage is negative, loss is positive (discourages action)
        policy_loss.append(-log_prob * advantage)
        
    policy_loss = torch.stack(policy_loss).sum()
    
    # Perform gradient descent to update the GNN weights
    optimizer.zero_grad()
    policy_loss.backward()
    optimizer.step()
    
    if episode % 10 == 0:
        print(f"Episode {episode:4d} | Total Reward: {total_reward:6.1f} | Baseline: {baseline_reward:6.1f} | Advantage: {advantage:6.1f}")

print("Training Complete!")
```

```python
policy_net.eval()

num_eval_attempts = 100
best_reward = -float('inf')
best_schedule = None

print(f"Generating {num_eval_attempts} schedules to find the best one...\n")

with torch.no_grad():
    for attempt in range(num_eval_attempts):
        env.reset()
        terminated = False
        truncated = False
        
        current_schedule_grid = np.full((10, 31), " - ", dtype=object)
        action_map = {0: " - ", 1: " M ", 2: " A "}
        current_reward = 0
        
        while not (terminated or truncated):
            g = convert_env_to_graph(env)
            action_mask = env.get_action_mask()
            
            emp_id = int(env.state[env.current_step, 0])
            day_id = int(env.state[env.current_step, 1])
            
            log_probs = policy_net(g, emp_id, day_id, action_mask)
            probs = torch.exp(log_probs)
            
            m = torch.distributions.Categorical(probs)
            action = m.sample().item()
            
            current_schedule_grid[emp_id, day_id] = action_map[action]
            
            _, reward, terminated, truncated, _ = env.step(action)
            current_reward += reward
            
        if current_reward > best_reward:
            best_reward = current_reward
            best_schedule = current_schedule_grid.copy()

print(f"Absolute Best Reward Found: {best_reward}\n")

header = "Emp ID | " + "".join([f"{d+1:3d}" for d in range(31)])
print(header)
print("-" * len(header))

for emp in range(10):
    row_string = f"Emp {emp:2d} | " + "".join(best_schedule[emp])
    print(row_string)
```
