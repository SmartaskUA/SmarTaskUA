```python
from typing import Optional
import datetime
import json
import numpy as np
import pandas as pd
import gymnasium as gym
from pathlib import Path

# ── Action encoding ──────────────────────────────────────────────────────────
# 0 = REST   1 = M-A   2 = T-A   3 = M-B   4 = T-B
ACTION_TO_SHIFT_TEAM = {1: ("M", "A"), 2: ("T", "A"), 3: ("M", "B"), 4: ("T", "B")}
ACTION_SHIFT_ORDER   = {0: 0, 1: 1, 2: 2, 3: 1, 4: 2}   # REST=0, Morning=1, Afternoon=2


def _build_special_days(year: int, n_days: int) -> set:
    """Return 0-based day indices that are Sundays or PT public holidays."""
    import holidays as hl
    pt_holidays = hl.country_holidays("PT", years=[year])
    start = datetime.date(year, 1, 1)
    special = set()
    for d in range(n_days):
        date = start + datetime.timedelta(days=d)
        if date.weekday() == 6 or date in pt_holidays:
            special.add(d)
    return special


class ScheduleEnv(gym.Env):
    """
    Gym environment for employee scheduling.

    Action space (one decision per employee-day step):
        0 = REST
        1 = Morning shift, Team A  (M-A)
        2 = Afternoon shift, Team A (T-A)
        3 = Morning shift, Team B  (M-B)
        4 = Afternoon shift, Team B (T-B)

    Hard constraints enforced via action masking:
        - Vacation blocking       : REST forced on vacation days
        - Annual day cap (≤ 223)  : max_days_per_year worked days
        - Consecutive day cap (≤ 5): max_consecutive_days in a row
        - Special days cap (≤ 22) : Sundays + PT public holidays
        - No backward shift       : Afternoon yesterday → no Morning today
        - Team eligibility        : only team-matching actions allowed

    Args:
        data_dir: path to SMARTASK_SIMPLE_2025 folder containing
                  problem.json, demand.csv, vacations.csv
    """

    NUM_ACTIONS = 5

    def __init__(self, data_dir: str = "../../../data/problems/SMARTASK_SIMPLE_2025"):
        super().__init__()
        base = Path(data_dir)

        # ── Load problem.json ────────────────────────────────────────────
        with open(base / "problem.json") as f:
            prob = json.load(f)

        self.num_days      = prob["temporalScope"]["numDays"]
        year               = prob["temporalScope"]["year"]
        employees          = prob["employees"]["simple"]
        self.num_employees = len(employees)

        # Team membership: list of sets per employee
        self.employee_teams = [set(emp.get("teams", [])) for emp in employees]

        # ── Load vacations.csv ───────────────────────────────────────────
        # Format: Employee Name, day1, day2, ..., day365   (1 = vacation)
        vac_df        = pd.read_csv(base / "vacations.csv", header=None)
        self.vac_mask = vac_df.iloc[:, 1:].values.astype(bool)   # (n_emp, n_days)

        # ── Load demand.csv ──────────────────────────────────────────────
        # Shape: (n_days, 2, 2)   shift: M=0 T=1   team: A=0 B=1
        dem_df            = pd.read_csv(base / "demand.csv")
        dem_df["date"]    = pd.to_datetime(dem_df["date"])
        start_ts          = pd.Timestamp(f"{year}-01-01")
        dem_df["day_idx"] = (dem_df["date"] - start_ts).dt.days

        self.min_demand = np.zeros((self.num_days, 2, 2), dtype=int)
        _shift_idx = {"M": 0, "T": 1}
        _team_idx  = {"A": 0, "B": 1}
        for _, row in dem_df.iterrows():
            d = int(row["day_idx"])
            s = _shift_idx[row["shift"]]
            t = _team_idx[row["team"]]
            self.min_demand[d, s, t] = int(row["minimum"])

        # ── Special days (Sundays + PT public holidays) ──────────────────
        self.special_days = _build_special_days(year, self.num_days)

        # ── Constraint parameters ────────────────────────────────────────
        self.max_days_per_year    = 223
        self.max_consecutive_days = 5
        self.special_days_cap     = 22

        # ── Gym spaces ───────────────────────────────────────────────────
        self.total_steps  = self.num_employees * self.num_days
        self.num_features = 2   # [Employee ID, Day ID]

        self.observation_space = gym.spaces.Box(
            low=0,
            high=max(self.num_employees, self.num_days),
            shape=(self.total_steps, self.num_features + self.NUM_ACTIONS),
            dtype=np.float32,
        )
        self.action_space = gym.spaces.Discrete(self.NUM_ACTIONS)

        self.reset()

    # ── Internal helpers ───────────────────────────────────────────────────

    def _build_initial_matrix(self) -> np.ndarray:
        matrix = np.zeros(
            (self.total_steps, self.num_features + self.NUM_ACTIONS), dtype=np.float32
        )
        row = 0
        for emp in range(self.num_employees):
            for day in range(self.num_days):
                matrix[row, 0] = emp
                matrix[row, 1] = day
                row += 1
        return matrix

    def _get_info(self) -> dict:
        return {}

    # ── Reward ─────────────────────────────────────────────────────────────

    def _calculate_reward(self, emp_id: int, day_id: int, action: int) -> float:                                                                                                                                                               
        reward = 0.0                                                                                                                                                                                                                           
    
        if action != 0:                                                                                                                                                                                                                        
            shift, team = ACTION_TO_SHIFT_TEAM[action]
            s = 0 if shift == "M" else 1
            t = 0 if team  == "A" else 1
            cov = self.daily_coverage[day_id, s, t]  # post-update                                                                                                                                                                             
            mn  = self.min_demand[day_id, s, t]
                                                                                                                                                                                                                                                
            if cov <= mn:                                                                                                                                                                                                                      
                # This unit filled a previously unmet demand slot → immediate reward
                reward += 2.0                                                                                                                                                                                                                  
            else:   
                # Slot was already covered → over-staffing penalty
                reward -= 2.0                                                                                                                                                                                                                  
        else:
            # REST chosen freely while behind pacing target                                                                                                                                                                                    
            target_rate = self.max_days_per_year / self.num_days
            expected    = int((day_id + 1) * target_rate)
            if self.days_worked[emp_id] < expected:                                                                                                                                                                                            
                reward -= 0.5
                                                                                                                                                                                                                                                
        return reward


    def _calculate_final_reward(self) -> float:                                                                                                                                                                                                
        reward = violations = 0.0                                                                                                                                                                                                              
        for day in range(self.num_days):
            for s in range(2):                                                                                                                                                                                                                 
                for t in range(2):                                                                                                                                                                                                             
                    shortfall = self.min_demand[day, s, t] - self.daily_coverage[day, s, t]
                    if shortfall > 0:                                                                                                                                                                                                          
                        reward -= shortfall * 5   # was 10 — per-step now carries this signal
                        violations += 1                                                                                                                                                                                                        
        for emp in range(self.num_employees):
            shortfall = self.max_days_per_year - self.days_worked[emp]                                                                                                                                                                         
            if shortfall > 0:
                violations += 1                                                                                                                                                                                                                
                reward -= shortfall * 10  # was 20
        if violations == 0:                                                                                                                                                                                                                    
            reward += 200  # was 400
        return reward   

    # ── Gym interface ──────────────────────────────────────────────────────

    def reset(self, seed: Optional[int] = None, options: Optional[dict] = None):
        super().reset(seed=seed)
        self.state        = self._build_initial_matrix()
        self.current_step = 0

        self.days_worked         = np.zeros(self.num_employees, dtype=int)
        self.consecutive_days    = np.zeros(self.num_employees, dtype=int)
        self.special_days_worked = np.zeros(self.num_employees, dtype=int)

        # Coverage: (n_days, 2, 2)  shift M=0 T=1, team A=0 B=1
        self.daily_coverage = np.zeros((self.num_days, 2, 2), dtype=np.float32)

        return self.state, self._get_info()

    def step(self, action: int):
        emp_id = int(self.state[self.current_step, 0])
        day_id = int(self.state[self.current_step, 1])

        self.state[self.current_step, 2 + action] = 1

        if action == 0:
            self.consecutive_days[emp_id] = 0
        else:
            self.days_worked[emp_id]      += 1
            self.consecutive_days[emp_id] += 1
            shift, team = ACTION_TO_SHIFT_TEAM[action]
            s = 0 if shift == "M" else 1
            t = 0 if team  == "A" else 1
            self.daily_coverage[day_id, s, t] += 1
            if day_id in self.special_days:
                self.special_days_worked[emp_id] += 1

        reward = self._calculate_reward(emp_id, day_id, action)

        self.current_step += 1
        terminated = self.current_step >= self.total_steps
        if terminated:
            reward += self._calculate_final_reward()

        return self.state.copy(), reward, terminated, False, self._get_info()

    def get_action_mask(self) -> np.ndarray:
        if self.current_step >= self.total_steps:
            return np.array([True, False, False, False, False])

        emp  = int(self.state[self.current_step, 0])
        day  = int(self.state[self.current_step, 1])
        mask = np.ones(self.NUM_ACTIONS, dtype=bool)

        # 1. Vacation day → REST only
        if self.vac_mask[emp, day]:
            mask[1:] = False
            return mask

        # 2. Annual day cap reached
        if self.days_worked[emp] >= self.max_days_per_year:
            mask[1:] = False

        # 3. Max consecutive days reached
        elif self.consecutive_days[emp] >= self.max_consecutive_days:
            mask[1:] = False

        else:
            # 4. Special day cap
            if day in self.special_days and self.special_days_worked[emp] >= self.special_days_cap:
                mask[1:] = False

            # 5. Backward shift: Afternoon yesterday → no Morning today
            if day > 0:
                y = emp * self.num_days + (day - 1)
                if self.state[y, 2 + 2] == 1 or self.state[y, 2 + 4] == 1:  # T-A or T-B
                    mask[1] = False   # M-A
                    mask[3] = False   # M-B

            # 6. Team eligibility
            for action, (_, team) in ACTION_TO_SHIFT_TEAM.items():
                if team not in self.employee_teams[emp]:
                    mask[action] = False

        # Safety: REST is always available
        if not np.any(mask):
            mask[0] = True

        return mask

    def render(self):
        labels = {0: "-", 1: "M-A", 2: "T-A", 3: "M-B", 4: "T-B"}
        for emp in range(self.num_employees):
            row = self.state[emp * self.num_days:(emp + 1) * self.num_days, 2:]
            schedule = [labels[int(np.argmax(r))] if r.sum() > 0 else "-" for r in row]
            print(f"Employee {emp + 1:2d}: {schedule}")
```

```python
import dgl
import torch
import numpy as np


def build_graph(env):
    sources      = env.state[:, 0].astype(np.int64)
    destinations = env.state[:, 1].astype(np.int64)
    graph_data = {
        ("employee", "assigned_to", "day"): (torch.tensor(sources),      torch.tensor(destinations)),
        ("day",      "staffed_by", "employee"): (torch.tensor(destinations), torch.tensor(sources)),
    }
    g = dgl.heterograph(graph_data)
    update_graph_features(g, env)
    return g


def update_graph_features(g, env):
    # Employee features: [days_worked, consecutive_days]
    emp_feat = np.column_stack((env.days_worked, env.consecutive_days))

    # Day features: coverage per (shift, team) — M-A, T-A, M-B, T-B
    day_feat = np.column_stack((
        env.daily_coverage[:, 0, 0],   # M-A
        env.daily_coverage[:, 1, 0],   # T-A
        env.daily_coverage[:, 0, 1],   # M-B
        env.daily_coverage[:, 1, 1],   # T-B
    ))

    g.nodes["employee"].data["feat"] = torch.tensor(emp_feat, dtype=torch.float32)
    g.nodes["day"].data["feat"]      = torch.tensor(day_feat, dtype=torch.float32)

    # Edge features: one-hot action assignment (5 actions)
    edge_attr = env.state[:, 2:2 + env.NUM_ACTIONS]
    g.edges["assigned_to"].data["feat"] = torch.tensor(edge_attr, dtype=torch.float32)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import dgl.nn as dglnn


class ActorCriticNetwork(nn.Module):
    """
    GNN-based Actor-Critic network for PPO + GAE.

    Shared GNN backbone encodes employee and day nodes.
    Actor head: outputs softmax action probabilities (with masking applied
                to logits before softmax, so masked actions get prob=0).
    Critic head: outputs scalar value V(s).

    Using probs (not logits) throughout — matches the reference PPO_discrete
    pattern and lets PyTorch's Categorical.entropy() handle zero-probability
    actions correctly (no 0 * -inf = nan).
    """

    def __init__(
        self,
        emp_in_feats: int = 2,
        day_in_feats: int = 4,
        hidden_dim:   int = 16,
        encoded_dim:  int = 64,
        num_actions:  int = 5,
        coverage_dim: int = 4,
        dyn_dim:      int = 2,
    ):
        super().__init__()

        # ── Shared GNN backbone ──────────────────────────────────────────
        self.emp_proj = nn.Linear(emp_in_feats, hidden_dim)
        self.day_proj = nn.Linear(day_in_feats, hidden_dim)

        self.conv1 = dglnn.HeteroGraphConv(
            {
                "assigned_to": dglnn.SAGEConv(hidden_dim, 32, "mean"),
                "staffed_by":  dglnn.SAGEConv(hidden_dim, 32, "mean"),
            },
            aggregate="sum",
        )
        self.conv2 = dglnn.HeteroGraphConv(
            {
                "assigned_to": dglnn.SAGEConv(32, encoded_dim, "mean"),
                "staffed_by":  dglnn.SAGEConv(32, encoded_dim, "mean"),
            },
            aggregate="sum",
        )

        ctx_dim = encoded_dim * 2 + coverage_dim + dyn_dim

        # ── Actor head ───────────────────────────────────────────────────
        self.actor_head = nn.Sequential(
            nn.Linear(ctx_dim, 64),
            nn.Tanh(),
            nn.Linear(64, num_actions),
        )

        # ── Critic head ──────────────────────────────────────────────────
        self.critic_head = nn.Sequential(
            nn.Linear(ctx_dim, 64),
            nn.Tanh(),
            nn.Linear(64, 1),
        )

    def encode(self, g):
        """Run GNN over the full heterograph. Returns (emp_embs [N,64], day_embs [D,64])."""
        h = {
            "employee": F.relu(self.emp_proj(g.nodes["employee"].data["feat"])),
            "day":      F.relu(self.day_proj(g.nodes["day"].data["feat"])),
        }
        h = self.conv1(g, h)
        h = {k: F.relu(v) for k, v in h.items()}
        h = {k: F.dropout(v, p=0.5, training=self.training) for k, v in h.items()}
        h = self.conv2(g, h)
        return h["employee"], h["day"]

    def _context(self, emp_emb, day_emb, coverage_gap, dyn_emp_feat):
        gap = torch.as_tensor(coverage_gap, dtype=torch.float32)
        dyn = torch.as_tensor(dyn_emp_feat, dtype=torch.float32)
        return torch.cat([emp_emb, day_emb, gap, dyn], dim=-1)

    def decode(self, emp_emb, day_emb, action_mask, coverage_gap, dyn_emp_feat):
        """
        Returns masked softmax probabilities for one step.

        Masking is applied to logits before softmax, so invalid actions
        get probability 0. If all actions are masked (shouldn't happen),
        REST (action 0) is forced valid as a fallback.
        """
        ctx    = self._context(emp_emb, day_emb, coverage_gap, dyn_emp_feat)
        logits = self.actor_head(ctx)
        mask   = torch.as_tensor(action_mask, dtype=torch.bool)
        if not mask.any():      # safety fallback: always allow REST
            mask[0] = True
        logits = logits.masked_fill(~mask, float("-inf"))
        return F.softmax(logits, dim=-1)  # probabilities, not log-probs

    def decode_value(self, emp_emb, day_emb, coverage_gap, dyn_emp_feat):
        """Returns scalar value estimate V(s)."""
        ctx = self._context(emp_emb, day_emb, coverage_gap, dyn_emp_feat)
        return self.critic_head(ctx).squeeze(-1)

    def forward(self, g, emp_id, day_id, action_mask, coverage_gap, dyn_emp_feat):
        emp_embs, day_embs = self.encode(g)
        return self.decode(emp_embs[emp_id], day_embs[day_id], action_mask, coverage_gap, dyn_emp_feat)

    def forward_from_stored(self, emp_embs_B, day_embs_B, cov_gaps_B, dyn_feats_B, masks_B):
        """
        Re-evaluate actor and critic on stored (detached) embeddings.
        Only actor_head and critic_head are called; no GNN forward pass.

        Returns probs (not log-probs) — caller creates Categorical(probs=...)
        to get entropy and log_prob cleanly, following the reference PPO_discrete.

        Args:
            emp_embs_B: [B, 64]
            day_embs_B: [B, 64]
            cov_gaps_B: [B, 4]
            dyn_feats_B: [B, 2]
            masks_B:    [B, 5] bool
        Returns:
            probs:  [B, 5]  softmax probabilities (masked)
            values: [B]     critic estimates
        """
        ctx    = torch.cat([emp_embs_B, day_embs_B, cov_gaps_B, dyn_feats_B], dim=-1)  # [B, 136]
        logits = self.actor_head(ctx)
        logits = logits.masked_fill(~masks_B, float("-inf"))
        probs  = F.softmax(logits, dim=-1)
        values = self.critic_head(ctx).squeeze(-1)
        return probs, values
```

```python
from dataclasses import dataclass, field
from torch.distributions import Categorical


@dataclass
class Trajectory:
    """Stores all data needed for a PPO update from one episode."""
    emp_embs:      list = field(default_factory=list)
    day_embs:      list = field(default_factory=list)
    coverage_gaps: list = field(default_factory=list)
    action_masks:  list = field(default_factory=list)
    actions:       list = field(default_factory=list)
    log_probs_old: list = field(default_factory=list)
    rewards:       list = field(default_factory=list)
    values:        list = field(default_factory=list)
    dyn_emp_feats: list = field(default_factory=list)

    def to_tensors(self):
        return {
            "emp_embs":      torch.stack(self.emp_embs),
            "day_embs":      torch.stack(self.day_embs),
            "coverage_gaps": torch.stack(self.coverage_gaps),
            "action_masks":  torch.stack(self.action_masks),
            "actions":       torch.tensor(self.actions,       dtype=torch.long),
            "log_probs_old": torch.tensor(self.log_probs_old, dtype=torch.float32),
            "rewards":       torch.tensor(self.rewards,       dtype=torch.float32),
            "values":        torch.tensor(self.values,        dtype=torch.float32),
            "dyn_emp_feats": torch.stack(self.dyn_emp_feats),
        }


def collect_trajectory(env, model, g):
    """
    Run one full episode and collect trajectory data for PPO.

    Follows reference PPO_discrete.choose_action:
      - actor returns softmax probs
      - Categorical(probs=...) for sampling and log_prob
    GNN re-run only when employee changes; embeddings stored detached.
    """
    model.eval()
    traj = Trajectory()

    env.reset()
    update_graph_features(g, env)

    current_emp   = -1
    emp_embs_full = None
    day_embs_full = None
    terminated = truncated = False

    with torch.no_grad():
        while not (terminated or truncated):
            emp_id = int(env.state[env.current_step, 0])
            day_id = int(env.state[env.current_step, 1])

            if emp_id != current_emp:
                update_graph_features(g, env)
                emp_embs_full, day_embs_full = model.encode(g)
                current_emp = emp_id

            emp_emb = emp_embs_full[emp_id].detach()
            day_emb = day_embs_full[day_id].detach()

            coverage_gap = [
                max(0, env.min_demand[day_id, 0, 0] - env.daily_coverage[day_id, 0, 0]),
                max(0, env.min_demand[day_id, 1, 0] - env.daily_coverage[day_id, 1, 0]),
                max(0, env.min_demand[day_id, 0, 1] - env.daily_coverage[day_id, 0, 1]),
                max(0, env.min_demand[day_id, 1, 1] - env.daily_coverage[day_id, 1, 1]),
            ]
            coverage_gap_t = torch.tensor(coverage_gap, dtype=torch.float32)

            action_mask   = env.get_action_mask()
            action_mask_t = torch.tensor(action_mask, dtype=torch.bool)

            dyn_emp_feat = [env.days_worked[emp_id], env.consecutive_days[emp_id]]
            dyn_emp_feat_t = torch.tensor(dyn_emp_feat, dtype=torch.float32)

            probs  = model.decode(emp_emb, day_emb, action_mask, coverage_gap, dyn_emp_feat)
            dist   = Categorical(probs=probs)
            action = dist.sample()
            value  = model.decode_value(emp_emb, day_emb, coverage_gap, dyn_emp_feat)

            _, reward, terminated, truncated, _ = env.step(action.item())

            traj.emp_embs.append(emp_emb)
            traj.day_embs.append(day_emb)
            traj.coverage_gaps.append(coverage_gap_t)
            traj.action_masks.append(action_mask_t)
            traj.actions.append(action.item())
            traj.log_probs_old.append(dist.log_prob(action).item())
            traj.rewards.append(float(reward))
            traj.values.append(value.item())
            traj.dyn_emp_feats.append(dyn_emp_feat_t)

    return traj
```

```python
def compute_gae(rewards, values, gamma=0.9995, lam=0.95):
    """
    Compute GAE advantages and critic targets (returns).

    A_t = sum_{k>=0} (gamma*lam)^k * delta_{t+k}
    delta_t = r_t + gamma*V(s_{t+1}) - V(s_t)

    Returns are computed from raw advantages BEFORE normalization,
    matching the reference: v_target = adv + vs (no_grad).
    Advantages are normalized after, for stable actor gradient scale.
    """
    T          = len(rewards)
    rewards_a  = np.array(rewards, dtype=np.float32)
    values_a   = np.array(values,  dtype=np.float32)
    values_ext = np.append(values_a, 0.0)  # V(s_T) = 0

    advantages = np.zeros(T, dtype=np.float32)
    gae        = 0.0
    for t in reversed(range(T)):
        delta         = rewards_a[t] + gamma * values_ext[t + 1] - values_ext[t]
        gae           = delta + gamma * lam * gae
        advantages[t] = gae

    returns    = advantages + values_a          # critic targets (before norm)
    advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-5)  # actor input

    return (
        torch.tensor(advantages, dtype=torch.float32),
        torch.tensor(returns,    dtype=torch.float32),
    )
```

```python
from torch.utils.data.sampler import BatchSampler, SubsetRandomSampler


def ppo_update(model, optimizer, batch, advantages, returns, clip_eps=0.2, value_coeff=0.5, entropy_coeff=0.01, K_epochs=4, mini_batch_size=512):
    T = batch["actions"].shape[0]
    losses = []
    track_entropy = []
    model.train()

    for _ in range(K_epochs):
        for index in BatchSampler(SubsetRandomSampler(range(T)), mini_batch_size, False):
            idx = torch.tensor(index)

            probs_new, values_new = model.forward_from_stored(
                batch["emp_embs"][idx],
                batch["day_embs"][idx],
                batch["coverage_gaps"][idx],
                batch["dyn_emp_feats"][idx], # Added in Fix 2
                batch["action_masks"][idx],
            )

            dist_now      = Categorical(probs=probs_new)
            a_logprob_now = dist_now.log_prob(batch["actions"][idx])
            dist_entropy  = dist_now.entropy()

            ratios = torch.exp(a_logprob_now - batch["log_probs_old"][idx])
            adv    = advantages[idx]
            surr1  = ratios * adv
            surr2  = torch.clamp(ratios, 1.0 - clip_eps, 1.0 + clip_eps) * adv
            actor_loss = -torch.min(surr1, surr2).mean() - entropy_coeff * dist_entropy.mean()

            # FIX: Use Huber Loss (smooth_l1_loss) on RAW returns
            critic_loss = F.smooth_l1_loss(values_new, returns[idx])

            loss = actor_loss + value_coeff * critic_loss

            track_entropy.append(dist_entropy.mean().item())

            optimizer.zero_grad()
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 0.5)
            optimizer.step()
            losses.append(loss.item())

    return float(np.mean(losses)), float(np.mean(track_entropy))
```

```python
import os

# ── Hyperparameters ──────────────────────────────────────────────────────────
NUM_EPISODES    = 10000
GAMMA           = 0.9995
LAM             = 0.95
CLIP_EPS        = 0.2
VALUE_COEFF     = 0.5
ENTROPY_COEFF_0 = 0.03
ENTROPY_MIN     = 0.02
K_EPOCHS        = 4
MINI_BATCH_SIZE = 512
LR              = 3e-4
SAVE_EVERY      = 50
RESUME_FROM     = "latest_ppo_scheduler.pth"

# ── Model & optimizer ────────────────────────────────────────────────────────
env       = ScheduleEnv()
model     = ActorCriticNetwork()
optimizer = torch.optim.Adam(model.parameters(), lr=LR, eps=1e-5)

best_reward   = -float("inf")
start_episode = 0

if os.path.exists(RESUME_FROM):
    ckpt = torch.load(RESUME_FROM, weights_only=True)
    model.load_state_dict(ckpt["model_state_dict"])
    optimizer.load_state_dict(ckpt["optimizer_state_dict"])
    start_episode = ckpt["episode"]
    best_reward   = float(ckpt.get("best_reward", -float("inf")))
    print(f"Resumed from episode {start_episode}, best reward = {best_reward:.1f}")

g = build_graph(env)

# ── Training loop ────────────────────────────────────────────────────────────
for episode in range(start_episode, NUM_EPISODES):
    entropy_coeff = max(ENTROPY_MIN, ENTROPY_COEFF_0 * (1.0 - episode / NUM_EPISODES))

    traj  = collect_trajectory(env, model, g)
    batch = traj.to_tensors()

    advantages, returns = compute_gae(traj.rewards, traj.values, GAMMA, LAM)

    mean_loss, mean_entropy = ppo_update(
        model, optimizer, batch, advantages, returns,
        clip_eps=CLIP_EPS,
        value_coeff=VALUE_COEFF,
        entropy_coeff=entropy_coeff,
        K_epochs=K_EPOCHS,
        mini_batch_size=MINI_BATCH_SIZE,
    )

    total_reward = sum(traj.rewards)

    if total_reward > best_reward:
        best_reward = float(total_reward)
        torch.save(
            {"episode": episode, "model_state_dict": model.state_dict(),
             "optimizer_state_dict": optimizer.state_dict(), "best_reward": best_reward},
            "best_ppo_scheduler.pth",
        )
        print(f"  NEW BEST: {best_reward:.1f} at episode {episode}")

    if episode % 10 == 0:
        print(f"Ep {episode:>4} | R={total_reward:>8.1f} | Best={best_reward:>8.1f} "
              f"| Loss={mean_loss:.4f} | ent={mean_entropy:.4f}")

    if (episode + 1) % SAVE_EVERY == 0:
        torch.save(
            {"episode": episode + 1, "model_state_dict": model.state_dict(),
             "optimizer_state_dict": optimizer.state_dict(), "best_reward": best_reward},
            "latest_ppo_scheduler.pth",
        )
```

```python
def evaluate(model, env, g, greedy=True):
    """Run one episode; greedy=True takes argmax, False samples from policy."""
    model.eval()
    env.reset()
    update_graph_features(g, env)

    current_emp = -1
    emp_embs_full = day_embs_full = None
    terminated = truncated = False
    total_reward = 0.0

    with torch.no_grad():
        while not (terminated or truncated):
            emp_id = int(env.state[env.current_step, 0])
            day_id = int(env.state[env.current_step, 1])
            if emp_id != current_emp:
                update_graph_features(g, env)
                emp_embs_full, day_embs_full = model.encode(g)
                current_emp = emp_id

            emp_emb = emp_embs_full[emp_id]
            day_emb = day_embs_full[day_id]
            coverage_gap = [
                max(0, env.min_demand[day_id, 0, 0] - env.daily_coverage[day_id, 0, 0]),
                max(0, env.min_demand[day_id, 1, 0] - env.daily_coverage[day_id, 1, 0]),
                max(0, env.min_demand[day_id, 0, 1] - env.daily_coverage[day_id, 0, 1]),
                max(0, env.min_demand[day_id, 1, 1] - env.daily_coverage[day_id, 1, 1]),
            ]
            action_mask = env.get_action_mask()
            probs  = model.decode(emp_emb, day_emb, action_mask, coverage_gap, dyn_emp_feat=[env.days_worked[emp_id], env.consecutive_days[emp_id]])
            action = probs.argmax() if greedy else Categorical(probs=probs).sample()
            _, reward, terminated, truncated, _ = env.step(action.item())
            total_reward += reward

    shortfall = int(np.maximum(0, env.min_demand - env.daily_coverage).sum())
    print(f"Total reward        : {total_reward:.1f}")
    print(f"Coverage shortfall  : {shortfall}")
    print(f"Days worked/employee: {env.days_worked.tolist()}")


ckpt = torch.load("best_ppo_scheduler.pth", weights_only=True)
model.load_state_dict(ckpt["model_state_dict"])
print(f"Checkpoint: episode {ckpt['episode']}, best reward = {ckpt['best_reward']:.1f}\n")
evaluate(model, env, g, greedy=False)
```
