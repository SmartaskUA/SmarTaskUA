```python
from typing import Optional
import numpy as np
import gymnasium as gym

class ScheduleEnv(gym.Env):
    def __init__(self, num_employees: int = 12, num_days: int = 365, num_shifts: int = 3):
        super(ScheduleEnv, self).__init__()
        self.num_employees = num_employees
        self.num_days = num_days
        self.num_shifts = num_shifts
        self.minimum_coverage = 2

        self.total_steps = num_employees * num_days
        self.num_features = 2  # [Employee ID, Day ID]

        self.max_days_per_month = 223
        self.max_consecutive_days = 5

        matrix_height = self.total_steps
        matrix_width = self.num_features + self.num_shifts
        self.observation_space = gym.spaces.Box(
            low=0,
            high=max(num_employees, num_days),
            shape=(matrix_height, matrix_width),
            dtype=np.float32
        )

        self.action_space = gym.spaces.Discrete(num_shifts)
        self.reset()

    def _build_initial_matrix(self):
        matrix = np.zeros((self.total_steps, self.num_features + self.num_shifts), dtype=np.float32)
        row_idx = 0
        for emp in range(self.num_employees):
            for day in range(self.num_days):
                matrix[row_idx, 0] = emp
                matrix[row_idx, 1] = day
                row_idx += 1
        return matrix

    def _get_info(self):
        return {}

    def _calculate_reward(self, emp_id: int, day_id: int, action: int):
        reward = 0

        if action == 0:
            if self.consecutive_days[emp_id] >= self.max_consecutive_days:
                reward += 5
        else:
            if action == 1 and self.daily_coverage_M[day_id] < self.minimum_coverage:
                reward += 4
            elif action == 2 and self.daily_coverage_T[day_id] < self.minimum_coverage:
                reward += 4

            if action == 1 and self.daily_coverage_M[day_id] >= self.minimum_coverage:
                reward -= 2
            elif action == 2 and self.daily_coverage_T[day_id] >= self.minimum_coverage:
                reward -= 2

            if self.days_worked[emp_id] <= self.max_days_per_month:
                reward += 1

        # Pacing reward — encourage working when behind schedule, resting when ahead
        days_per_day_target = self.max_days_per_month / self.num_days  # ~0.611
        expected_days = int((day_id + 1) * days_per_day_target)
        if action != 0 and self.days_worked[emp_id] < expected_days:
            reward += 0.5   # behind pace, good to work
        elif action == 0 and self.days_worked[emp_id] >= expected_days:
            reward += 0.3   # on/ahead of pace, good to rest

        return reward

    def _calculate_final_reward(self):
        reward = 0

        violations = 0
        for day in range(self.num_days):
            if self.daily_coverage_M[day] < self.minimum_coverage:
                reward -= (self.minimum_coverage - self.daily_coverage_M[day]) * 10
                violations += 1
            if self.daily_coverage_T[day] < self.minimum_coverage:
                reward -= (self.minimum_coverage - self.daily_coverage_T[day]) * 10
                violations += 1

        for emp in range(self.num_employees):
            shortfall = self.max_days_per_month - self.days_worked[emp]
            if shortfall > 0:
                violations += 1
                reward -= shortfall * 20

        if violations == 0:
            reward += 400

        return reward

    def reset(self, seed: Optional[int] = None, options: Optional[dict] = None):
        super().reset(seed=seed)
        self.state = self._build_initial_matrix()
        self.current_step = 0

        self.days_worked = np.zeros(self.num_employees, dtype=int)
        self.consecutive_days = np.zeros(self.num_employees, dtype=int)

        self.daily_coverage_T = np.zeros(self.num_days, dtype=np.float32)
        self.daily_coverage_M = np.zeros(self.num_days, dtype=np.float32)

        return self.state, self._get_info()

    def step(self, action: int):
        emp_id = int(self.state[self.current_step, 0])
        day_id = int(self.state[self.current_step, 1])

        self.state[self.current_step, 2 + action] = 1

        if action == 0:
            self.consecutive_days[emp_id] = 0
        else:
            self.days_worked[emp_id] += 1
            self.consecutive_days[emp_id] += 1

            if action == 1:
                self.daily_coverage_M[day_id] += 1
            elif action == 2:
                self.daily_coverage_T[day_id] += 1

        reward = self._calculate_reward(emp_id, day_id, action)

        self.current_step += 1

        terminated = self.current_step >= self.total_steps
        if terminated:
            reward += self._calculate_final_reward()

        truncated = False
        return self.state.copy(), reward, terminated, truncated, self._get_info()

    def render(self):
        print("Current Schedule State:")
        for emp in range(self.num_employees):
            emp_schedule = self.state[emp * self.num_days:(emp + 1) * self.num_days, 2:]
            print(f"Employee {emp}: {emp_schedule}")

    def get_action_mask(self):
        if self.current_step >= self.total_steps:
            return np.array([True, False, False])

        emp = int(self.state[self.current_step, 0])
        day = int(self.state[self.current_step, 1])

        mask = np.ones(3, dtype=bool)

        # Max days reached — only rest allowed
        if self.days_worked[emp] >= self.max_days_per_month:
            mask[1] = False
            mask[2] = False

        # 5-consecutive-day constraint — always enforced (hard constraint)
        if self.consecutive_days[emp] >= self.max_consecutive_days:
            mask[1] = False
            mask[2] = False

        # Backward shift constraint — Afternoon yesterday blocks Morning today
        if day > 0:
            yesterday_step = (emp * self.num_days) + (day - 1)
            if self.state[yesterday_step, 2 + 2] == 1:
                mask[1] = False

        if not np.any(mask):
            mask[0] = True

        return mask
```

```python
import dgl
import torch
import numpy as np

def build_graph(env):
    sources = env.state[:, 0].astype(np.int64)     
    destinations = env.state[:, 1].astype(np.int64)
    
    graph_data = {
        ('employee', 'assigned_to', 'day'): (torch.tensor(sources), torch.tensor(destinations)),
        ('day', 'staffed_by', 'employee'): (torch.tensor(destinations), torch.tensor(sources))
    }
    g = dgl.heterograph(graph_data)
    update_graph_features(g, env)
    return g

def update_graph_features(g, env):
    emp_feat = np.column_stack((env.days_worked, env.consecutive_days))
    day_feat = np.column_stack((env.daily_coverage_M, env.daily_coverage_T))
    g.nodes['employee'].data['feat'] = torch.tensor(emp_feat, dtype=torch.float32)
    g.nodes['day'].data['feat'] = torch.tensor(day_feat, dtype=torch.float32)
    
    edge_attr = env.state[:, 2:5]
    g.edges['assigned_to'].data['feat'] = torch.tensor(edge_attr, dtype=torch.float32)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import dgl.nn as dglnn

class PolicyNetwork(nn.Module):
    def __init__(self, emp_in_feats=2, day_in_feats=2, hidden_dim=16, encoded_dim=64, num_actions=3):
        super(PolicyNetwork, self).__init__()
        
        self.emp_proj = nn.Linear(emp_in_feats, hidden_dim)
        self.day_proj = nn.Linear(day_in_feats, hidden_dim)
        
        self.conv1 = dglnn.HeteroGraphConv({
            'assigned_to': dglnn.SAGEConv(hidden_dim, 32, 'mean'),
            'staffed_by': dglnn.SAGEConv(hidden_dim, 32, 'mean')
        }, aggregate='sum')
        
        self.conv2 = dglnn.HeteroGraphConv({
            'assigned_to': dglnn.SAGEConv(32, encoded_dim, 'mean'),
            'staffed_by': dglnn.SAGEConv(32, encoded_dim, 'mean')
        }, aggregate='sum')
        
        self.scoring_layer = nn.Sequential(
            nn.Linear(encoded_dim * 2 + 2, 64),
            nn.ReLU(),
            nn.Linear(64, num_actions)
        )

    def encode(self, g):
        h = {
            'employee': F.relu(self.emp_proj(g.nodes['employee'].data['feat'])),
            'day':      F.relu(self.day_proj(g.nodes['day'].data['feat']))
        }
        h = self.conv1(g, h)
        h = {k: F.relu(v) for k, v in h.items()}
        h = {k: F.dropout(v, p=0.5, training=self.training) for k, v in h.items()}
        h = self.conv2(g, h)
        return h['employee'], h['day']

    def decode(self, emp_emb, day_emb, action_mask, coverage_gap):
        gap_tensor = torch.tensor(coverage_gap, dtype=torch.float32)
        combined = torch.cat([emp_emb, day_emb, gap_tensor], dim=-1)
        logits = self.scoring_layer(combined)
        
        mask_tensor = torch.tensor(action_mask, dtype=torch.bool)
        logits = logits.masked_fill(~mask_tensor, float('-inf'))
        
        return F.log_softmax(logits, dim=-1)

    def forward(self, g, current_emp_id, current_day_id, action_mask, coverage_gap):
        emp_embs, day_embs = self.encode(g)
        return self.decode(emp_embs[current_emp_id], day_embs[current_day_id], action_mask, coverage_gap)
```

```python
import torch
import torch.optim as optim
from torch.distributions import Categorical
import numpy as np
import os

env = ScheduleEnv()
policy_net = PolicyNetwork(emp_in_feats=2, day_in_feats=2, hidden_dim=16, encoded_dim=64, num_actions=3)
optimizer = optim.Adam(policy_net.parameters(), lr=1e-4)

num_episodes = 5000
gamma = 1.0
entropy_coeff = 0.05
baseline_alpha = 0.05
save_every = 50

# --- Resume from latest checkpoint if available ---
RESUME_FROM = '4k-episodes.pth'

start_episode = 0
best_total_reward = -float('inf')
baseline = 0.0

if os.path.exists(RESUME_FROM):
    checkpoint = torch.load(RESUME_FROM, weights_only=True)
    policy_net.load_state_dict(checkpoint['model_state_dict'])
    optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
    start_episode = checkpoint['episode']
    print(f"Resumed from '{RESUME_FROM}' at episode {start_episode}")
    if os.path.exists('best_scheduler.pth'):
        best_ckpt = torch.load('best_scheduler.pth', weights_only=False)
        best_total_reward = float(best_ckpt.get('best_reward', -float('inf')))
        baseline = best_total_reward
        print(f"Best reward carried over: {best_total_reward}")
else:
    print("No checkpoint found, starting from scratch.")

print("Starting GNN Training...")

for episode in range(start_episode, num_episodes):
    env.reset()

    saved_log_probs = []
    saved_probs = []
    rewards = []

    terminated = False
    truncated = False

    g = build_graph(env)

    current_emp = -1
    emp_embs = None
    day_embs = None

    entropy_coeff = 0.05 * max(0.2, 1.0 - episode / 5000)

    while not (terminated or truncated):
        emp_id = int(env.state[env.current_step, 0])
        day_id = int(env.state[env.current_step, 1])

        coverage_gap = [
            max(0, env.minimum_coverage - env.daily_coverage_M[day_id]),
            max(0, env.minimum_coverage - env.daily_coverage_T[day_id])
        ]

        if emp_id != current_emp:
            update_graph_features(g, env)
            emp_embs, day_embs = policy_net.encode(g)
            current_emp = emp_id

        action_mask = env.get_action_mask()
        log_probs = policy_net.decode(emp_embs[emp_id], day_embs[day_id], action_mask, coverage_gap)
        probs = torch.exp(log_probs)

        m = Categorical(probs)
        action = m.sample()

        saved_log_probs.append(m.log_prob(action))
        saved_probs.append(probs)

        _, reward, terminated, truncated, _ = env.step(action.item())
        rewards.append(reward)

    total_reward_val = sum(rewards)

    # Update EMA baseline before computing returns
    baseline = (1 - baseline_alpha) * baseline + baseline_alpha * total_reward_val

    # Loss calculation
    returns = []
    R = 0
    for r in rewards[::-1]:
        R = r + gamma * R
        returns.insert(0, R)
    returns = torch.tensor(returns)
    returns = returns - baseline
    returns = (returns - returns.mean()) / (returns.std() + 1e-9)

    policy_loss = []
    entropy_loss = 0
    for log_prob, prob, R in zip(saved_log_probs, saved_probs, returns):
        policy_loss.append(-log_prob * R)
        entropy_loss += -torch.sum(prob * torch.log(prob + 1e-9))

    loss = torch.stack(policy_loss).sum() - entropy_coeff * entropy_loss

    optimizer.zero_grad()
    loss.backward()
    torch.nn.utils.clip_grad_norm_(policy_net.parameters(), max_norm=1.0)
    optimizer.step()

    # Save best checkpoint — store as float to avoid numpy scalar in pickle
    if total_reward_val > best_total_reward:
        best_total_reward = float(total_reward_val)
        print(f"NEW HIGH SCORE: {best_total_reward} at Episode {episode:>4} (Model Saved)")
        torch.save({
            'episode': episode,
            'model_state_dict': policy_net.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'best_reward': best_total_reward
        }, 'best_scheduler.pth')

    if episode % 10 == 0:
        print(f"Episode {episode:>4} | Total Reward: {total_reward_val} | Best Reward: {best_total_reward}")

    # Periodic checkpoint so progress isn't lost if execution is interrupted
    if (episode + 1) % save_every == 0:
        torch.save({
            'episode': episode + 1,
            'model_state_dict': policy_net.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
        }, 'latest_scheduler.pth')

torch.save({
    'episode': num_episodes,
    'model_state_dict': policy_net.state_dict(),
    'optimizer_state_dict': optimizer.state_dict(),
}, 'latest_scheduler.pth')
```

```python
import torch
import numpy as np

policy_net = PolicyNetwork(emp_in_feats=2, day_in_feats=2, hidden_dim=16, encoded_dim=64, num_actions=3)
checkpoint = torch.load('4k-episodes.pth', weights_only=True)
policy_net.load_state_dict(checkpoint['model_state_dict'])
policy_net.eval()

print(f"Loaded checkpoint from episode {checkpoint['episode']}")
action_map = {0: " - ", 1: " M ", 2: " A "}


def run_episode(temperature: float = 1.0):
    env.reset()
    terminated = False
    truncated = False
    schedule_grid = np.full((env.num_employees, env.num_days), " - ", dtype=object)
    total_reward = 0

    g = build_graph(env)
    current_emp = -1
    emp_embs = None
    day_embs = None

    while not (terminated or truncated):
        emp_id = int(env.state[env.current_step, 0])
        day_id = int(env.state[env.current_step, 1])

        coverage_gap = [
            max(0, env.minimum_coverage - env.daily_coverage_M[day_id]),
            max(0, env.minimum_coverage - env.daily_coverage_T[day_id])
        ]

        if emp_id != current_emp:
            update_graph_features(g, env)
            emp_embs, day_embs = policy_net.encode(g)
            current_emp = emp_id

        action_mask = env.get_action_mask()
        log_probs = policy_net.decode(emp_embs[emp_id], day_embs[day_id], action_mask, coverage_gap)
        # probs = torch.exp(log_probs)
        # print(f"Emp {emp_id}, Day {day_id} | Action Mask: {action_mask} | Probs: {probs.tolist()}")

        if temperature == 0.0:
            action = torch.argmax(log_probs).item()
        else:
            scaled_log_probs = log_probs / temperature
            # re-normalise after temperature scaling (mask already applied in decode)
            scaled_log_probs = scaled_log_probs - torch.logsumexp(scaled_log_probs, dim=0)
            action = torch.distributions.Categorical(torch.exp(scaled_log_probs)).sample().item()

        schedule_grid[emp_id, day_id] = action_map[action]
        _, reward, terminated, truncated, _ = env.step(action)
        total_reward += reward

    violations = sum(
        (env.minimum_coverage - env.daily_coverage_M[day]) for day in range(env.num_days)
        if env.daily_coverage_M[day] < env.minimum_coverage
    )

    violations += sum(
        (env.minimum_coverage - env.daily_coverage_T[day]) for day in range(env.num_days)
        if env.daily_coverage_T[day] < env.minimum_coverage
    )

    return schedule_grid, total_reward, violations, env.days_worked.copy()


header = "Emp ID | " + "".join([f"{d+1:3d}" for d in range(env.num_days)])

# --- Standard stochastic runs (T=1.0, 20 attempts) ---
print("\n=== Stochastic Sampling T=1.0 (20 attempts) ===")
with torch.no_grad():
    best_reward = -float('inf')
    best_grid = None
    best_violations = -1
    best_days = None
    second_best_grid = None
    second_best_days = None

    for attempt in range(20):
        grid, reward, violations, days_worked = run_episode(temperature=1.0)
        print(f"  Attempt {attempt+1:2d} | Reward: {reward:8.1f} | Violations: {violations}")
        if reward > best_reward:
            best_reward = reward
            best_grid = grid.copy()
            best_violations = violations
            best_days = days_worked.copy()
        if violations < best_violations:
            second_best_grid = grid.copy()
            second_best_days = days_worked.copy()

print(f"\nBest T=1.0 reward: {best_reward:.1f} | Violations: {best_violations}")
print(f"Days worked per employee: {best_days.tolist()}  (target: 223 each)")
print("\n" + header)
print("-" * len(header))
for emp in range(env.num_employees):
    print(f"Emp {emp:2d} | " + "".join(best_grid[emp]))


print("--------------------------------")
print(f"Days worked per employee in second-best (lowest violations) schedule: {second_best_days.tolist()}  (target: 223 each)")
for emp in range(env.num_employees):
    print(f"Emp {emp:2d} | " + "".join(second_best_grid[emp]))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Run one stochastic episode with the current policy
env_check = ScheduleEnv()
obs, _ = env_check.reset()
checkpoint = torch.load('best_scheduler.pth', weights_only=False)
policy_net.load_state_dict(checkpoint['model_state_dict'])
policy_net.eval()

with torch.no_grad():
    g = build_graph(env_check)
    done = False
    while not done:
        update_graph_features(g, env_check)
        emp_id = int(env_check.state[env_check.current_step, 0])
        day_id = int(env_check.state[env_check.current_step, 1])
        action_mask = env_check.get_action_mask()
        coverage_gap = [
            max(0, env_check.minimum_coverage - env_check.daily_coverage_M[day_id]),
            max(0, env_check.minimum_coverage - env_check.daily_coverage_T[day_id])
        ]
        log_probs = policy_net(g, emp_id, day_id, action_mask, coverage_gap)
        action = torch.distributions.Categorical(torch.exp(log_probs)).sample().item()
        obs, reward, terminated, truncated, _ = env_check.step(action)
        done = terminated or truncated

# Plot coverage by day
days = np.arange(env_check.num_days)
plt.figure(figsize=(14, 4))
plt.plot(days, env_check.daily_coverage_M, label='Morning', alpha=0.7)
plt.plot(days, env_check.daily_coverage_T, label='Afternoon', alpha=0.7)
plt.axhline(y=2, color='r', linestyle='--', label='Minimum (2)')
plt.xlabel('Day of year')
plt.ylabel('Employees assigned')
plt.title('Coverage distribution across the year')
plt.legend()
plt.tight_layout()
plt.show()

# Summary stats by quarter
for q, (start, end) in enumerate([(0,91),(91,182),(182,273),(273,365)], 1):
    m = env_check.daily_coverage_M[start:end]
    t = env_check.daily_coverage_T[start:end]
    viol_m = np.sum(m < 2)
    viol_t = np.sum(t < 2)
    print(f"Q{q} (days {start}-{end}): Morning violations={viol_m}, Afternoon violations={viol_t}, avg_M={m.mean():.2f}, avg_T={t.mean():.2f}")
```
